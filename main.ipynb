```python
import numpy as np
import ot

# Initialize 1D histograms a and b (must sum to 1 and be positive)
a = np.array([0.5, 0.3, 0.2])  # 3-element distribution
b = np.array([0.2, 0.5, 0.3])  # 3-element distribution

# Initialize ground cost matrix M (3x3)
M = np.array([[0., 1., 2.],
              [1., 0., 1.],
              [2., 1., 0.]])

# Regularization parameter
reg = 0.1
```

### Compute Wasserstein distances

```python
# a,b are 1D histograms (sum to 1 and positive)
# M is the ground cost matrix

# With the unified  API :
Wd = ot.solve(M, a, b).value                # exact linear program
Wd_reg = ot.solve(M, a, b, reg=reg).value   # entropic regularized OT

# Print
print(Wd)
print(Wd_reg)
```

### Compute OT matrix

```python
# a,b are 1D histograms (sum to 1 and positive)
# M is the ground cost matrix

# With the unified API :
T = ot.solve(M, a, b).plan # exact linear program
T_reg = ot.solve(M, a, b, reg=reg).plan # entropic regularized 

# Print
print(T)
print(T_reg)
```

```python
# Initialize X and Y as 2D arrays of shape (n_samples, n_features)
np.random.seed(42)
n_samples_X = 5
n_samples_Y = 4
n_features = 2

X = np.random.randn(n_samples_X, n_features)  # 5 samples, 2 features
Y = np.random.randn(n_samples_Y, n_features)  # 4 samples, 2 features

print("X shape:", X.shape)
print("Y shape:", Y.shape)
print(X)
print(Y)
```

### Compute OT on empirical distributions

```python
# X and Y are two 2D arrays of shape (n_samples, n_features)

# with squared euclidean metric
T = ot.solve_sample(X, Y).plan # exact linear program
T_reg = ot.solve_sample(X, Y, reg=reg).plan # entropic regularized OT

print(T)
print(T_reg)

Wass_2 = ot.solve_sample(X, Y).value # Squared Wasserstein_2
Wass_1 = ot.solve_sample(X, Y, metric='euclidean').value # Wasserstein 1

print("Squared Wasserstein-2:", Wass_2)
print("Wasserstein-1:", Wass_1)
```

### Compute Wasserstein barycenter

```python
# Initialize A as an n*d matrix containing d 1D histograms
# Each row is a 1D histogram that sums to 1
n_histograms = 3
n_bins = 4

A = np.array([[0.5, 0.3, 0.2, 0.0],
              [0.2, 0.3, 0.3, 0.2],
              [0.1, 0.2, 0.3, 0.4]])  # n=3 histograms, d=4 bins each

# Initialize ground cost matrix M (4x4 for the 4 bins)
M = np.array([[0., 1., 2., 3.],
              [1., 0., 1., 2.],
              [2., 1., 0., 1.],
              [3., 2., 1., 0.]])

print("\nM (cost matrix):\n", M)
```

```python
# A is a n*d matrix containing d  1D histograms
# M is the ground cost matrix
reg = 0.1
ba = ot.barycenter(A.T, M, reg) # reg is regularization parameter

print("Barycenter:", ba)
```
